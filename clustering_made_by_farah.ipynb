```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

def run_kmeans_on_subset(df_subset, features, n_clusters=5):
    # 1. Scaling
    scaler = StandardScaler()
    df_scaled = scaler.fit_transform(df_subset[features])

    # 2. Clustering
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(df_scaled)

    # 3. Evaluation
    silhouette = silhouette_score(df_scaled, labels)
    db_index = davies_bouldin_score(df_scaled, labels)

    return labels, silhouette, db_index
```

```python
features = ['gender_encoded', 'nationality_encoded', 'age', 'hard_skills', 'soft_skills', 'creativity', 'teamwork']

# Large subset
_, sil_large, db_large = run_kmeans_on_subset(df_subset_large, features)

# Small subset
_, sil_small, db_small = run_kmeans_on_subset(df_subset_small, features)

# Class-specific subset
_, sil_class, db_class = run_kmeans_on_subset(df_subset_class, features)

# Show results
print("📊 KMeans Evaluation Summary")
print(f"Large Subset (3000):     Silhouette = {sil_large:.4f}, DB Index = {db_large:.4f}")
print(f"Small Subset (500):      Silhouette = {sil_small:.4f}, DB Index = {db_small:.4f}")
print(f"Class-Specific Subset:   Silhouette = {sil_class:.4f}, DB Index = {db_class:.4f}")
```

```python
import numpy as np
import pandas as pd
from collections import defaultdict
from scipy.stats import entropy

def form_groups_from_clusters(df, cluster_labels, group_size=5):
    df = df.copy()
    df['cluster'] = cluster_labels

    # Create a dictionary: cluster -> list of student indices
    clusters = defaultdict(list)
    for idx, c in enumerate(cluster_labels):
        clusters[c].append(idx)

    # Shuffle indices within each cluster for randomness
    for c in clusters:
        np.random.shuffle(clusters[c])

    groups = []
    group = []

    # Distribute students from different clusters into groups round-robin
    while True:
        added = False
        for c in clusters:
            if clusters[c]:
                group.append(clusters[c].pop())
                added = True
                if len(group) == group_size:
                    groups.append(group)
                    group = []
        if not added:  # no students left
            if group:  # add last group even if smaller
                groups.append(group)
            break

    return groups
```

```python
def calculate_entropy_for_groups(df, groups, column_name):
    entropies = []
    for group in groups:
        values = df.iloc[group][column_name]
        counts = values.value_counts(normalize=True)
        ent = entropy(counts, base=2)
        entropies.append(ent)
    return np.mean(entropies)
```

```python
from sklearn.cluster import KMeans
import numpy as np
import pandas as pd
from collections import defaultdict
from scipy.stats import entropy

def form_groups_from_clusters(df, cluster_labels, group_size=5):
    df = df.copy()
    df['cluster'] = cluster_labels
    clusters = defaultdict(list)
    for idx, c in enumerate(cluster_labels):
        clusters[c].append(idx)
    for c in clusters:
        np.random.shuffle(clusters[c])
    groups = []
    group = []
    while True:
        added = False
        for c in clusters:
            if clusters[c]:
                group.append(clusters[c].pop())
                added = True
                if len(group) == group_size:
                    groups.append(group)
                    group = []
        if not added:
            if group:
                groups.append(group)
            break
    return groups

def calculate_entropy_for_groups(df, groups, column_name):
    entropies = []
    for group in groups:
        values = df.iloc[group][column_name]
        counts = values.value_counts(normalize=True)
        ent = entropy(counts, base=2)
        entropies.append(ent)
    return np.mean(entropies)

def run_kmeans_and_evaluate_entropy(df_subset, features, group_size=5, n_clusters=5):
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(df_subset[features])
    groups = form_groups_from_clusters(df_subset, labels, group_size=group_size)
    gender_ent = calculate_entropy_for_groups(df_subset, groups, 'gender')
    nationality_ent = calculate_entropy_for_groups(df_subset, groups, 'nationality')
    return gender_ent, nationality_ent

# Features list (make sure these are properly prepared in your subsets)
features = ['gender_encoded', 'nationality_encoded', 'age', 'hard_skills', 'soft_skills', 'creativity', 'teamwork']

# Run and print for all subsets:
for name, subset in [('Large Subset (3000)', df_subset_large),
                     ('Small Subset (500)', df_subset_small),
                     ('Class-Specific Subset', df_subset_class)]:
    gender_entropy, nationality_entropy = run_kmeans_and_evaluate_entropy(subset, features)
    print(f"{name} - Gender Entropy: {gender_entropy:.4f}, Nationality Entropy: {nationality_entropy:.4f}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import AgglomerativeClustering

def run_agglomerative_cosine(df_subset, features, n_clusters=5):
    from sklearn.preprocessing import StandardScaler
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(df_subset[features])

    # Compute cosine distance (1 - cosine similarity)
    cosine_dist = 1 - cosine_similarity(X_scaled)

    model = AgglomerativeClustering(n_clusters=n_clusters, metric='precomputed', linkage='average')
    labels = model.fit_predict(cosine_dist)

    sil_score, db_score = evaluate_clustering(X_scaled, labels)
    return labels, sil_score, db_score
```

```python
from scipy.stats import entropy

def compute_group_entropy(df, labels, col1, col2):
    df = df.copy()
    df['cluster'] = labels
    entropies_col1 = []
    entropies_col2 = []

    for cluster_id in df['cluster'].unique():
        cluster_data = df[df['cluster'] == cluster_id]

        # Compute normalized value counts
        dist_col1 = cluster_data[col1].value_counts(normalize=True)
        dist_col2 = cluster_data[col2].value_counts(normalize=True)

        # Compute entropy
        ent_col1 = entropy(dist_col1, base=2)
        ent_col2 = entropy(dist_col2, base=2)

        entropies_col1.append(ent_col1)
        entropies_col2.append(ent_col2)

    # Return average entropy per cluster
    return sum(entropies_col1)/len(entropies_col1), sum(entropies_col2)/len(entropies_col2)
```

```python
# Run Agglomerative with Cosine on all subsets
labels_large, sil_large, db_large = run_agglomerative_cosine(df_subset_large, features)
labels_small, sil_small, db_small = run_agglomerative_cosine(df_subset_small, features)
labels_class, sil_class, db_class = run_agglomerative_cosine(df_subset_class, features)

# Group Entropy
ent_gender_large, ent_nat_large = compute_group_entropy(df_subset_large, labels_large, 'gender_encoded', 'nationality_encoded')
ent_gender_small, ent_nat_small = compute_group_entropy(df_subset_small, labels_small, 'gender_encoded', 'nationality_encoded')
ent_gender_class, ent_nat_class = compute_group_entropy(df_subset_class, labels_class, 'gender_encoded', 'nationality_encoded')

# Display Results
print("\n📊 Agglomerative Clustering (Cosine) – Evaluation Summary")
print(f"Large Subset (3000):     Silhouette = {sil_large:.4f}, DB Index = {db_large:.4f}")
print(f"Small Subset (500):      Silhouette = {sil_small:.4f}, DB Index = {db_small:.4f}")
print(f"Class-Specific Subset:   Silhouette = {sil_class:.4f}, DB Index = {db_class:.4f}")

print("\n📊 Agglomerative Clustering - Group Diversity (Entropy)")
print(f"Large Subset (3000) - Gender Entropy: {ent_gender_large:.4f}, Nationality Entropy: {ent_nat_large:.4f}")
print(f"Small Subset (500)  - Gender Entropy: {ent_gender_small:.4f}, Nationality Entropy: {ent_nat_small:.4f}")
print(f"Class-Specific Subset - Gender Entropy: {ent_gender_class:.4f}, Nationality Entropy: {ent_nat_class:.4f}")
```

```python
import gower
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.stats import entropy

# Evaluate clustering (adjusted for distance matrix input)
def evaluate_clustering(X_or_dist, labels, is_distance_matrix=False):
    if is_distance_matrix:
        sil = silhouette_score(X_or_dist, labels, metric='precomputed')
        db = -1  # DB index not valid for precomputed dist
    else:
        from sklearn.metrics import davies_bouldin_score
        sil = silhouette_score(X_or_dist, labels)
        db = davies_bouldin_score(X_or_dist, labels)
    return sil, db

# Compute entropy for gender and nationality
def compute_group_entropy(df, labels, col1, col2):
    df = df.copy()
    df['cluster'] = labels
    entropies_col1 = []
    entropies_col2 = []

    for cluster_id in df['cluster'].unique():
        cluster_data = df[df['cluster'] == cluster_id]
        dist_col1 = cluster_data[col1].value_counts(normalize=True)
        dist_col2 = cluster_data[col2].value_counts(normalize=True)
        ent_col1 = entropy(dist_col1, base=2)
        ent_col2 = entropy(dist_col2, base=2)
        entropies_col1.append(ent_col1)
        entropies_col2.append(ent_col2)

    return sum(entropies_col1)/len(entropies_col1), sum(entropies_col2)/len(entropies_col2)

# Run Agglomerative clustering with Gower distance
def run_agglomerative_gower(df_subset, features, n_clusters=5):
    # Compute Gower distance matrix
    gower_dist = gower.gower_matrix(df_subset[features])

    # Agglomerative clustering with precomputed Gower distance
    model = AgglomerativeClustering(n_clusters=n_clusters, metric='precomputed', linkage='average')
    labels = model.fit_predict(gower_dist)

    # Evaluate (silhouette using distance matrix)
    sil_score, db_score = evaluate_clustering(gower_dist, labels, is_distance_matrix=True)
    return labels, sil_score, db_score

# Features to use
features = ['gender_encoded', 'nationality_encoded', 'age', 'hard_skills', 'soft_skills', 'creativity', 'teamwork']

# Run on all subsets
labels_large, sil_large, db_large = run_agglomerative_gower(df_subset_large, features)
labels_small, sil_small, db_small = run_agglomerative_gower(df_subset_small, features)
labels_class, sil_class, db_class = run_agglomerative_gower(df_subset_class, features)

# Compute entropy scores
ent_gender_large, ent_nat_large = compute_group_entropy(df_subset_large, labels_large, 'gender_encoded', 'nationality_encoded')
ent_gender_small, ent_nat_small = compute_group_entropy(df_subset_small, labels_small, 'gender_encoded', 'nationality_encoded')
ent_gender_class, ent_nat_class = compute_group_entropy(df_subset_class, labels_class, 'gender_encoded', 'nationality_encoded')

# Print results
print("📊 Agglomerative Clustering (Gower) – Evaluation Summary")
print(f"Large Subset (3000):     Silhouette = {sil_large:.4f}, DB Index = {db_large}")
print(f"Small Subset (500):      Silhouette = {sil_small:.4f}, DB Index = {db_small}")
print(f"Class-Specific Subset:   Silhouette = {sil_class:.4f}, DB Index = {db_class}")

print("\n📊 Agglomerative Clustering - Group Diversity (Entropy)")
print(f"Large Subset (3000) - Gender Entropy: {ent_gender_large:.4f}, Nationality Entropy: {ent_nat_large:.4f}")
print(f"Small Subset (500)  - Gender Entropy: {ent_gender_small:.4f}, Nationality Entropy: {ent_nat_small:.4f}")
print(f"Class-Specific Subset - Gender Entropy: {ent_gender_class:.4f}, Nationality Entropy: {ent_nat_class:.4f}")
```

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import numpy as np

def run_dbscan_on_subset(df_subset, features, eps=0.5, min_samples=5):
    # Extract features and scale if not done already
    X = df_subset[features].values

    # Run DBSCAN
    dbscan = DBSCAN(eps=eps, min_samples=min_samples, metric='euclidean')
    labels = dbscan.fit_predict(X)

    # Filter out noise points (-1) for evaluation
    mask = labels != -1
    if np.sum(mask) > 1:  # Need at least 2 clusters
        sil_score = silhouette_score(X[mask], labels[mask])
    else:
        sil_score = -1  # Not enough clusters for Silhouette

    # Davies-Bouldin score requires at least 2 clusters
    unique_labels = set(labels)
    if len(unique_labels) > 1 and -1 in unique_labels:
        unique_labels.remove(-1)
    if len(unique_labels) > 1:
        from sklearn.metrics import davies_bouldin_score
        db_score = davies_bouldin_score(X[mask], labels[mask])
    else:
        db_score = -1  # Not valid

    return labels, sil_score, db_score

# Run on subsets
features = ['gender_encoded', 'nationality_encoded', 'age', 'hard_skills', 'soft_skills', 'creativity', 'teamwork']

labels_large, sil_large, db_large = run_dbscan_on_subset(df_subset_large, features, eps=1.5, min_samples=3)
labels_small, sil_small, db_small = run_dbscan_on_subset(df_subset_small, features, eps=1.5, min_samples=3)
labels_class, sil_class, db_class = run_dbscan_on_subset(df_subset_class, features, eps=1.5, min_samples=3)

from scipy.stats import entropy

def compute_entropy_ignore_noise(df, labels, col1, col2):
    df = df.copy()
    df['cluster'] = labels
    df = df[df['cluster'] != -1]  # Remove noise points

    unique_clusters = df['cluster'].unique()
    if len(unique_clusters) == 0:
        return 0.0, 0.0  # No clusters found, return zero entropy

    entropies_col1 = []
    entropies_col2 = []

    for cluster_id in unique_clusters:
        cluster_data = df[df['cluster'] == cluster_id]

        dist_col1 = cluster_data[col1].value_counts(normalize=True)
        dist_col2 = cluster_data[col2].value_counts(normalize=True)

        ent_col1 = entropy(dist_col1, base=2)
        ent_col2 = entropy(dist_col2, base=2)

        entropies_col1.append(ent_col1)
        entropies_col2.append(ent_col2)

    avg_entropy_col1 = sum(entropies_col1) / len(entropies_col1)
    avg_entropy_col2 = sum(entropies_col2) / len(entropies_col2)

    return avg_entropy_col1, avg_entropy_col2



ent_gender_large, ent_nat_large = compute_entropy_ignore_noise(df_subset_large, labels_large, 'gender_encoded', 'nationality_encoded')
ent_gender_small, ent_nat_small = compute_entropy_ignore_noise(df_subset_small, labels_small, 'gender_encoded', 'nationality_encoded')
ent_gender_class, ent_nat_class = compute_entropy_ignore_noise(df_subset_class, labels_class, 'gender_encoded', 'nationality_encoded')

print("📊 DBSCAN Evaluation Summary")
print(f"Large Subset (3000):     Silhouette = {sil_large:.4f}, DB Index = {db_large:.4f}")
print(f"Small Subset (500):      Silhouette = {sil_small:.4f}, DB Index = {db_small:.4f}")
print(f"Class-Specific Subset:   Silhouette = {sil_class:.4f}, DB Index = {db_class:.4f}")

print("\n📊 DBSCAN - Group Diversity (Entropy)")
print(f"Large Subset (3000) - Gender Entropy: {ent_gender_large:.4f}, Nationality Entropy: {ent_nat_large:.4f}")
print(f"Small Subset (500)  - Gender Entropy: {ent_gender_small:.4f}, Nationality Entropy: {ent_nat_small:.4f}")
print(f"Class-Specific Subset - Gender Entropy: {ent_gender_class:.4f}, Nationality Entropy: {ent_nat_class:.4f}")
```

```python
from minisom import MiniSom
import numpy as np

def run_minisom_on_subset(df_subset, features, som_x=10, som_y=10, sigma=1.0, learning_rate=0.5, num_iterations=1000):
    X = df_subset[features].values

    som = MiniSom(som_x, som_y, X.shape[1], sigma=sigma, learning_rate=learning_rate, random_seed=42)
    som.train_random(X, num_iterations)

    # Map each sample to its BMU (Best Matching Unit)
    bmu_indices = np.array([som.winner(x) for x in X])
    # Flatten BMU indices into cluster labels
    labels = bmu_indices[:,0] * som_y + bmu_indices[:,1]

    sil_score, db_score = evaluate_clustering(X, labels)

    return labels, sil_score, db_score

# Run on all subsets
labels_large, sil_large, db_large = run_minisom_on_subset(df_subset_large, features)
labels_small, sil_small, db_small = run_minisom_on_subset(df_subset_small, features)
labels_class, sil_class, db_class = run_minisom_on_subset(df_subset_class, features)

# Compute entropy for group diversity
ent_gender_large, ent_nat_large = compute_group_entropy(df_subset_large, labels_large, 'gender_encoded', 'nationality_encoded')
ent_gender_small, ent_nat_small = compute_group_entropy(df_subset_small, labels_small, 'gender_encoded', 'nationality_encoded')
ent_gender_class, ent_nat_class = compute_group_entropy(df_subset_class, labels_class, 'gender_encoded', 'nationality_encoded')

print("📊 MiniSom Evaluation Summary")
print(f"Large Subset (3000):     Silhouette = {sil_large:.4f}, DB Index = {db_large:.4f}")
print(f"Small Subset (500):      Silhouette = {sil_small:.4f}, DB Index = {db_small:.4f}")
print(f"Class-Specific Subset:   Silhouette = {sil_class:.4f}, DB Index = {db_class:.4f}")

print("\n📊 MiniSom - Group Diversity (Entropy)")
print(f"Large Subset (3000) - Gender Entropy: {ent_gender_large:.4f}, Nationality Entropy: {ent_nat_large:.4f}")
print(f"Small Subset (500)  - Gender Entropy: {ent_gender_small:.4f}, Nationality Entropy: {ent_nat_small:.4f}")
print(f"Class-Specific Subset - Gender Entropy: {ent_gender_class:.4f}, Nationality Entropy: {ent_nat_class:.4f}")
```
